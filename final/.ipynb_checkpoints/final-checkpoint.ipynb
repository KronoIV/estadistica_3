```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
import re
import seaborn as sns

df = pd.read_csv('trainTw.csv') 

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)  
    return text.split()

all_words = []
df['texto'].apply(lambda x: all_words.extend(preprocess_text(x)))
word_freq = Counter(all_words)
common_words = word_freq.most_common(10)


words, counts = zip(*common_words)
plt.figure(figsize=(10, 6))
sns.barplot(x=list(counts), y=list(words))
plt.title("10 palabras más frecuentes en los tweets")
plt.xlabel("Frecuencia")
plt.ylabel("Palabra")
plt.grid(axis='x')
plt.tight_layout()
plt.show()

correlation_matrix = df.corr(numeric_only=True)
label_correlation = correlation_matrix["label"].drop("label")
print("Correlación de la variable 'label':")
print(label_correlation)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer
from nltk.corpus import stopwords
import nltk
from nltk.tokenize import word_tokenize
from collections import Counter

# Descargar recursos necesarios de NLTK
nltk.download('stopwords')
nltk.download('punkt')

# Cargar los datos
data = pd.read_csv('trainTw.csv')

# Filtrar solo los registros con is_train = TRUE
train_data = data[data['is_train'] == True]

# 1. Gráfico de barras con las 10 palabras más frecuentes (solo datos de entrenamiento)

# Preparar el texto: unir todos los tweets de entrenamiento y limpiar
all_text = ' '.join(train_data['texto'].str.lower())

# Tokenizar y eliminar stopwords en español
stop_words = set(stopwords.words('spanish'))
words = word_tokenize(all_text)
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]

# Contar frecuencias
word_counts = Counter(filtered_words)
top_10_words = word_counts.most_common(10)

# Crear gráfico
plt.figure(figsize=(10, 6))
plt.bar([word[0] for word in top_10_words], [word[1] for word in top_10_words])
plt.title('Top 10 palabras más frecuentes en tweets de entrenamiento')
plt.xlabel('Palabra')
plt.ylabel('Frecuencia')
plt.xticks(rotation=45)
plt.show()

# 2. Análisis de correlación con la variable objetivo (solo datos de entrenamiento)

# Crear matriz de términos del documento
vectorizer = CountVectorizer(stop_words=list(stop_words))
X = vectorizer.fit_transform(train_data['texto'])

# Convertir a DataFrame para análisis
words_df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())

# Añadir la variable objetivo
words_df['label'] = train_data['label']

# Calcular correlaciones
correlations = words_df.corr()['label'].sort_values(ascending=False)

# Mostrar las 10 palabras con mayor correlación positiva y negativa
print("\nTop 10 palabras con mayor correlación positiva (aumento del dólar):")
print(correlations[1:11])  # Excluimos la correlación consigo misma

print("\nTop 10 palabras con mayor correlación negativa (fortalecimiento del peso):")
print(correlations[-10:])
```
