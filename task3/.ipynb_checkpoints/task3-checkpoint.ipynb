```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import datasets, layers, models, preprocessing
import pandas as pd
import seaborn as sns
import os
import matplotlib.pyplot as plt
from PIL import Image
import sklearn as sk
import csv
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from tensorflow.keras import layers


train_path = os.fsencode("train")
test_path = os.fsencode("test")
```

```python
#Modelo con regresion logistica
```

```python
train_dataset = tf.keras.utils.image_dataset_from_directory(
    directory=train_path.decode(),
    validation_split=0.2,
    subset="training",
    seed=4014,
    image_size=(64, 64)
)

val_dataset = tf.keras.utils.image_dataset_from_directory(
    directory=train_path.decode(),
    validation_split=0.2,
    subset="validation",
    seed=4014,
    image_size=(64, 64)
)
def preprocess_dataset(ds):
    X, y = [], []
    for images, labels in ds:
        X.append(tf.reshape(images, [images.shape[0], -1]).numpy())
        y.append(labels.numpy())
    return np.concatenate(X), np.concatenate(y)

X_train, y_train = preprocess_dataset(train_dataset)
X_val, y_val = preprocess_dataset(val_dataset)

model = LogisticRegression(penalty='l2', max_iter=500, C=1.0, random_state=42)
model.fit(X_train, y_train)

train_preds = model.predict(X_train)
val_preds = model.predict(X_val)

print("Accuracy en entrenamiento:", accuracy_score(y_train, train_preds))
print("Accuracy en validación:", accuracy_score(y_val, val_preds))


def predict_batch_images(test_dir):
    processed_imgs = []
    file_names = []
    
    for filename in os.listdir(test_dir):
        path = os.path.join("test", filename.decode())
        img = tf.keras.utils.load_img(path, target_size=(64, 64))
        img_arr = tf.keras.utils.img_to_array(img)
        processed_imgs.append(img_arr)
        file_names.append(filename.decode())

    img_tensor = np.array(processed_imgs)
    flat_imgs = img_tensor.reshape((img_tensor.shape[0], -1))

    predictions = model.predict(flat_imgs)
    label_names = [train_dataset.class_names[idx] for idx in predictions]

    return file_names, label_names

file_list, label_list = predict_batch_images(test_path)

submission = pd.DataFrame({
    "file": file_list,
    "label": label_list
})
submission.to_csv("logistica.csv", index=False)
```

```python
train_data = tf.keras.utils.image_dataset_from_directory(
    directory=train_path.decode(),
    validation_split=0.2,
    subset="training",
    seed=7527,
    image_size=(64, 64)
)

val_data = tf.keras.utils.image_dataset_from_directory(
    directory=train_path.decode(),
    validation_split=0.2,
    subset="validation",
    seed=7527,
    image_size=(64, 64)
)

plt.figure(figsize=(12, 12))
for img_batch, label_batch in train_data.take(1):
    for idx in range(25):
        plt.subplot(5, 5, idx + 1)
        plt.imshow(img_batch[idx].numpy().astype("uint8"))
        plt.title(train_data.class_names[label_batch[idx]])
        plt.axis("off")

model = models.Sequential([
    layers.Input(shape=(64, 64, 3)),

    layers.Conv2D(32, kernel_size=(3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Conv2D(64, kernel_size=(3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Conv2D(128, kernel_size=(3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),

    layers.Dense(128, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.25),

    layers.Dense(16, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy']
)

stop_early = tf.keras.callbacks.EarlyStopping(
    monitor='loss',
    patience=5,
    restore_best_weights=True
)

# Training
model.fit(
    train_data,
    validation_data=val_data,
    epochs=10,
    callbacks=[stop_early]
)
```

```python
def generate_predictions_cnn(image_dir):
    image_data = []
    filenames = []

    for file in os.listdir(image_dir):
        path = os.path.join("test", file.decode())
        img = tf.keras.utils.load_img(path, target_size=(64, 64))
        img_array = tf.keras.utils.img_to_array(img)
        image_data.append(img_array)
        filenames.append(file.decode())

    input_batch = np.array(image_data)
    raw_output = model.predict(input_batch)
    predicted_classes = np.argmax(raw_output, axis=1)
    label_names = [train_data.class_names[i] for i in predicted_classes]

    return filenames, label_names

file_list, label_list = generate_predictions_cnn(test_path)

submission_df = pd.DataFrame({
    "file": file_list,
    "label": label_list
})
submission_df.to_csv("cnn.csv", index=False)
```
